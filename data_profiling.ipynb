```python
chicago = spark.table("bronze_chicago")
dallas = spark.table("bronze_dallas")
```

```python
print("Chicago Count:", chicago.count())
print("Dallas Count:", dallas.count())

chicago.printSchema()
dallas.printSchema()
```

```python
from pyspark.sql.functions import col, sum

def null_counts(df):
    return df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns])

null_counts(chicago).show()
null_counts(dallas).show()
```

```python
fact_inspections = spark.table("fact_inspections")
dim_restaurant = spark.table("dim_restaurant")
dim_location = spark.table("dim_location")
```

```python
fact_step1 = fact_inspections.join(
    dim_restaurant,
    on="business_name",
    how="left"
)
```

```python
fact_step2 = fact_step1.join(
    dim_location,
    on=["city", "zip_code"],
    how="left"
)
```

```python
fact_final = fact_step2.select(
    "inspection_id",
    "inspection_date",
    "inspection_score",
    "inspection_result",
    "restaurant_id",
    "location_id"
)
```

```python
fact_final.write.mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("fact_inspections_final")
```

```python
from pyspark.sql.functions import upper
from pyspark.sql.functions import col
dim_location = dim_location.withColumn("city", upper(col("city")))
```

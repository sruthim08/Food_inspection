```python
final_violations = spark.table("fact_violation")
```

```python
dim_violation = final_violations.select("violation").distinct()
```

```python
from pyspark.sql.functions import monotonically_increasing_id

dim_violation = dim_violation.withColumn(
    "violation_id",
    monotonically_increasing_id()
)
```

```python
dim_violation = dim_violation.select("violation_id", "violation")
```

```python
dim_violation.write.mode("overwrite").saveAsTable("dim_violation")
```

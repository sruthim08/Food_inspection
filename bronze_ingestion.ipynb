```python
chicago_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/workspace/default/dadabi_spring2026/Food_Inspections_20260410.csv")

dallas_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/workspace/default/dadabi_spring2026/Restaurant_and_Food_Establishment_Inspections_(October_2016_to_January_2024)_20260410.csv")
```

```python
def clean_column_names(df):
    new_cols = []
    for col_name in df.columns:
        clean_name = col_name.lower() \
            .replace(" ", "_") \
            .replace("#", "num") \
            .replace("(", "") \
            .replace(")", "")
        new_cols.append(clean_name)
    return df.toDF(*new_cols)

chicago_df = clean_column_names(chicago_df)
dallas_df = clean_column_names(dallas_df)
```

```python
chicago_df.write.mode("overwrite").saveAsTable("bronze_chicago")
dallas_df.write.mode("overwrite").saveAsTable("bronze_dallas")
```

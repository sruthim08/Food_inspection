```python
chicago = spark.table("silver_chicago")
dallas = spark.table("silver_dallas")
```

```python
from pyspark.sql.functions import col, lit, monotonically_increasing_id
chicago_fact = chicago.select(
    "inspection_id",
    "inspection_date",
    "inspection_score",
    "inspection_result",
    "business_name",
    "zip_code",
    "city"
    )
```

```python
from pyspark.sql.functions import lit
from pyspark.sql.functions import col

dallas_fact = dallas.select(
    "inspection_id",
    "inspection_date",
    "inspection_score",
    "business_name",
    "zip_code"
).withColumn(
    "inspection_result",
    col("inspection_score").cast("string")   # ✅ FIX HERE
).withColumn(
    "city",
    lit("Dallas")
)
```

```python
fact_inspections = chicago_fact.unionByName(dallas_fact)
```

```python
fact_inspections.write.mode("overwrite").saveAsTable("fact_inspections")
```
